# 4.Planned KPIs

The dashboard will display several key performance indicators to summarise the machine learning model performance and overall churn results.


### Total Customers | Shows the total number of customers included in eachd ataset
### Churn Rate | Shows the percentage of customers who left the service or bank
### Best Model | Displays the selected best performing model for each dataset based on evaluation metrics.
### Accuracy | Measures the overall percentage of correct predictions made by the model.
### Precision | Measures how many predicted churn customers were actual churn customers
### Recall | Measures how well the model identified actual churn customers.
### F1-Score | Provides a balance between precision and recall.
### ROC-AUC | Measures the model's ability to distinguish between churn and non churn customers.

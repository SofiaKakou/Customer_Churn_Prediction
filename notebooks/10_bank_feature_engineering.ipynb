# 7.Feature Selection
The final feature set is selected by separating the input variables from the target ones.
- x contains the input features used to make predictions
- y contains the churn target variable

```python
# define features and target
x = df.drop("churn", axis=1)
y = df["churn"]

print("X shape:", x.shape)
print("y shape:", y.shape)
```

```python
x.head()
```

```python
y.head()
```

# 8.Train-Test Split
The dataset is split into training and testing sets.

- 80% of the data is used for training the ML
- 20% of the data is used for testing

A fixed random_state is used to make the split reproducible.

```python
# Creating the Split
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
    )

print("X Train Shape:", x_train.shape)
print("y Train Shape:", y_train.shape)
print("X Test Shape:", x_test.shape)
print("y Test Shape:", y_test.shape)
```

```python
print("Training target distribution:")
print(y_train.value_counts(normalize=True)*100)

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True)*100)
```

# Observations
The target variable is moderately imbalanced with approximately 80% non churn and 20% churn cases. The consistent distribution across training and testing sets ensues a reliable model evaluation, although appropriate metrics must be use to account for class imbalance.

# 9.Scale Numerical Features

The scaling happens so that features with larger numeric ranges do not dominate models such as Logistic Regression.

Scaling is applied after the train/ test split to avoid leakage.

The scaler is fitted only on the training data, then applied to both training and testing data.

```python
# identifying numerical cols
num_cols = x_train.select_dtypes(include=["int64", "float64"]).columns.tolist()

num_cols
```

```python
# scale numerical features
scaler = StandardScaler()

x_train_scaled = x_train.copy()
x_test_scaled = x_test.copy()

x_train_scaled[num_cols] = scaler.fit_transform(x_train[num_cols])
x_test_scaled[num_cols] = scaler.transform(x_test[num_cols])
```

```python
# scaled training data preview
x_train_scaled.head()
```

# 10.Save Prepared Data

```python
save_folder = "/content/drive/MyDrive/Thesis/Prepared Data/"

x_train_scaled.to_csv(save_folder + "bank_x_train.csv", index=False)
x_test_scaled.to_csv(save_folder + "bank_x_test.csv", index=False)
y_train.to_csv(save_folder + "bank_y_train.csv", index=False)
y_test.to_csv(save_folder + "bank_y_test.csv", index=False)
```
